## 1. Le problème des listes

Convertir des prix en dollars avec une liste demande une boucle. Le taux de 131 gourdes pour un dollar est un taux d'exemple.

```python
prix = [488, 512, 530, 547]
en_dollars = []
for p in prix:
    en_dollars.append(round(p / 131, 2))
en_dollars
```

```python
prix / 131
```

Une liste ne sait pas se diviser par un nombre : `TypeError`. NumPy règle ce problème.

## 2. NumPy : calculer sur tout le tableau d'un coup

```python
import numpy as np

tableau = np.array([488, 512, 530, 547])
tableau / 131
```

```python
np.round(tableau / 131, 2)
```

```python
tableau + 10
```

```python
tableau.mean(), tableau.max(), tableau.min()
```

Une comparaison donne un tableau de booléens. `sum` compte les `True`.

```python
tableau > 520
```

```python
(tableau > 520).sum()
```

## 3. La Series : un tableau avec des étiquettes

Une Series de pandas, c'est un tableau NumPy dont chaque valeur a une étiquette : l'**index**.

```python
import pandas as pd

riz = pd.Series([488, 512, 530], index=["janvier", "février", "mars"])
riz
```

```python
riz["février"]
```

```python
riz.mean()
```
